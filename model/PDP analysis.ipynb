```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV files
csv_day_7 = pd.read_csv('pdp_results_7.csv')  # Change the file names as necessary
csv_day_28 = pd.read_csv('pdp_results_28.csv')
csv_day_56 = pd.read_csv('pdp_results_56.csv')

# Set up the figure
# plt.figure(figsize=(12, 8))

# Plot XGBoost predictions for day 7
plt.plot(csv_day_7['FA% (Original Scale)'].values, csv_day_7['XGBoost'].values, label='AGE=7', color='blue')

# Plot XGBoost predictions for day 28
plt.plot(csv_day_28['FA% (Original Scale)'].values, csv_day_28['XGBoost'].values, label='AGE=28', color='orange')

# Plot XGBoost predictions for day 56
plt.plot(csv_day_56['FA% (Original Scale)'].values, csv_day_56['XGBoost'].values, label='AGE=56', color='green')

# Adding titles and labels
# plt.title('Relationship Between FA% and XGBoost Predictions Across Different Days')
plt.xlabel('FA% ')
plt.ylabel('Predicted fc')
plt.legend()
plt.grid()
# plt.show()
plt.xlim(0,0.3)
plt.tight_layout()

# Save the figure with better resolution and spacing
plt.savefig('PDP_XGBoost.png', dpi=300)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV file
df = pd.read_excel(r'updated_fc_predictions_with_fc_hat.xlsx')  # Update this path to your CSV file

colors = ['blue', 'orange', 'green']

# Plot histograms for each AGE category
for age, color in zip([7, 28, 56], colors):
    # Select Fc values for the specific AGE
    subset = df[df['AGE'] == age]
    
    # Plot the histogram with density=True for normalization
    plt.hist(subset['fc (MPa)'], bins=20, alpha=0.5, color=color, label=f'AGE: {age} Days', density=True)

# Adding titles and labels
# plt.title('Distribution of Fc Values Across Different Ages')
plt.xlabel('fc (MPa)')
plt.ylabel('Density')
plt.legend()
plt.grid()

# Set y-axis limits
# plt.ylim(0, 1)

# Show the plot
# plt.show()
plt.tight_layout()

# Save the figure with better resolution and spacing
plt.savefig('density by age.png', dpi=300)
```
