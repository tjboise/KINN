```python
import pandas as pd
# import xgboost as xgb
from BorutaShap import BorutaShap

# Load your dataset
filename = 'normal.xlsx'
sheetname = 'normal'
df = pd.read_excel(filename, sheet_name=sheetname, header=0)

# Define features and target variable
X = df[['AGE','PC', 'PC_TYPE', 'FA', 'SS', 'SF', 'FAGG', 'CAGG', 'WATER', 'AEA', 'WR_HR', 'WR', 'ACC', 
        'VOID', 'w/b', 'b/a', 'CAGG%', 'FAGG%','FA%','SS%','SF%']]
y = df['fc (MPa)']

# Initialize a model
# Uncomment the following line if using XGBoost
# model = xgb.XGBRegressor()

# Initialize BorutaSHAP with the model
Feature_Selector = BorutaShap(importance_measure='shap', classification=False)

# Fit BorutaSHAP to the data
Feature_Selector.fit(X=X, y=y, n_trials=100, sample=False, train_or_test='test', normalize=True, verbose=True)

# Get the selected features
selected_features = Feature_Selector.Subset()
print("Selected features:", selected_features)

# Plot feature importance
Feature_Selector.plot(which_features='all')
```
