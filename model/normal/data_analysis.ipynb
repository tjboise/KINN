```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the specific sheet 'age56' from the uploaded Excel file
file_path = 'normal.xlsx'
df_age56 = pd.read_excel(file_path, sheet_name='age56')

# Plot a boxplot with x-axis as 'PC_TYPE' and y-axis as 'fc (MPa)'
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_age56, x='PC_TYPE', y='fc (MPa)')
plt.title('Box Plot of fc (MPa) by PC_TYPE for Age 56')
plt.xlabel('PC_TYPE')
plt.ylabel('fc (MPa)')
plt.xticks(rotation=45)
plt.show()
```
