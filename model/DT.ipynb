```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import matplotlib.pyplot as plt

# Load and preprocess the data
df = pd.read_excel(r'../data.xlsx', sheet_name='Sheet10')
df.dropna(inplace=True)

# Define features and target
X = df[['AGE', 'PC', 'FA', 'FAGG', 'CAGG', 'WATER', 'AEA', 'WR_HR', 'WR', 'ACC',  'LATEX',
       'SLUMP', 'VOID', 'TOTAL_BINDER', 'w/b', 'b/a', 'SCM%']].values  # Add all relevant features
y = df['fc (MPa)'].values



# Standardize the features (optional, Decision Trees don't need this but it helps with better understanding)
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Split the data into training and test sets
Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, train_size=0.85, random_state=50)

# Create the Decision Tree model with some tuning
tree_model = DecisionTreeRegressor(max_depth=15, min_samples_split=15, random_state=50)

# Train the Decision Tree model
tree_model.fit(Xtrain, ytrain)

# Predict on the training and testing datasets
ytrain_pred = tree_model.predict(Xtrain)
ytest_pred = tree_model.predict(Xtest)

# Evaluate the model
train_r2 = r2_score(ytrain, ytrain_pred)
test_r2 = r2_score(ytest, ytest_pred)

print(f'Train R-squared: {train_r2}')
print(f'Test R-squared: {test_r2}')

# Calculate RMSE and MAE
R2 = r2_score(ytest, ytest_pred)
RMSE = mean_squared_error(ytest, ytest_pred) ** 0.5
MAE = mean_absolute_error(ytest, ytest_pred)

print(f'R-squared is {R2}, RMSE is {RMSE}, and MAE is {MAE}.')

# # ---- Error Distribution Plot ----
# errors = ytest - ytest_pred

# plt.figure(figsize=(8, 6))
# plt.hist(errors, bins=30, alpha=0.7, color='blue')
# plt.title('Error Distribution-decision tree')
# plt.xlabel('Error (fc_actual - fc_predicted)', fontsize=12)
# plt.ylabel('Frequency', fontsize=12)
# plt.grid(True)
# plt.show()

# # ---- Scatter Plot of Actual vs Predicted fc ----
# plt.figure(figsize=(8, 6))

# # Scatter plot of actual vs predicted fc values
# plt.scatter(ytest, ytest_pred, color='blue', alpha=0.6, label='Predicted vs Actual')

# # Plot the diagonal line representing perfect predictions (x = y)
# plt.plot([min(ytest), max(ytest)], [min(ytest), max(ytest)], color='red', linestyle='--', label='Perfect Prediction')

# plt.xlabel('Actual fc (MPa)', fontsize=12)
# plt.ylabel('Predicted fc (MPa)', fontsize=12)
# plt.title('Actual vs Predicted Compressive Strength-decision tree', fontsize=14)
# plt.legend()
# plt.grid(True)
# plt.show()
```
