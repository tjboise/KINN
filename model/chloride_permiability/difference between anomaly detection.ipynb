```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_excel(r'cleaned_dataset_with_iso_forest.xlsx', sheet_name='Sheet1')

# List of features for box plots
features = ['AGE', 'PC1', 'PC2', 'PC1_TYPE', 'FA', 'SS', 'SF', 'FAGG', 'CAGG', 'WATER', 
            'AEA', 'WR_HR', 'WR', 'ACC', 'VOID', 'FIBER', 'SLUMP', 'IS_PUMP', 'IS_LATEX', 
            'LATEX_AMOUNT', 'IS_LIGHT_WEIGHT', 'APPLICATION_ID', 'TOTAL_BINDER', 'TOTAL_AGG', 
            'w/b', 'b/a', 'SCM%']

# Create box plots for all the selected features
plt.figure(figsize=(15, 10))
df[features].boxplot()

# Rotate x-axis labels for better readability
plt.xticks(rotation=90)
plt.title("Box Plot of Selected Features")
plt.ylabel("Values")

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the datasets
df = pd.read_excel(r'cleaned_dataset_with_iso_forest.xlsx', sheet_name='Ori Cleaned Data')
df2 = pd.read_excel(r'cleaned_dataset_with_iso_forest.xlsx', sheet_name='Sheet1')

# Add a column to indicate the dataset source
df['Dataset'] = 'Original Data'
df2['Dataset'] = 'After Anomaly Detection'

# Combine both datasets
df_combined = pd.concat([df, df2])

# List of features for box plots
features = ['AGE', 'PC1', 'FA', 'SS', 'FAGG', 'CAGG', 'WATER', 
             'WR_HR',  'TOTAL_BINDER', 'TOTAL_AGG']

# Melt the dataframe to long format for seaborn boxplot
df_melted = df_combined.melt(id_vars='Dataset', value_vars=features, var_name='Feature', value_name='Value')

# Create a box plot using seaborn
# plt.figure(figsize=(15, 10))
sns.boxplot(x='Feature', y='Value', hue='Dataset', data=df_melted)

# Rotate x-axis labels for better readability
plt.xticks(rotation=90)
# plt.title("Box Plot of Selected Features with Normal vs Anomaly")
plt.ylabel("Values")
# plt.yscale('log')

# Show the plot
# plt.tight_layout()
# plt.show()
plt.savefig('boxplot1.png', dpi=300, bbox_inches='tight')
```
