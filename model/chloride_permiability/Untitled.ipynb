```python
from autogluon.tabular import TabularDataset, TabularPredictor
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import numpy as np

# Load the dataset
filename = 'cleaned_dataset_with_iso_forest.xlsx'
sheetname = 'Sheet1'
dataset = pd.read_excel(filename, sheetname, header=0)

# Split the data into train and test sets
train_data, test_data = train_test_split(dataset, test_size=0.2, random_state=42)

# Label column
label = 'log'

# Train the model using a regression preset (no need to drop the label here)
predictor = TabularPredictor(label=label, problem_type='regression').fit(
    train_data.drop(columns=['COULOMB_TEST', 'FA%', 'SS%', 'SF%', 'Chloride Ion Penetrability Class']), 
    presets='best_quality'
)

# Get a list of all model names from the leaderboard
all_models = predictor.get_model_names()

# Collect metrics for each model
metrics = []
for model_name in all_models:
    # Drop the label column and irrelevant columns from the test data
    y_pred = predictor.predict(test_data.drop(columns=[label, 'COULOMB_TEST', 'FA%', 'SS%', 'SF%', 'Chloride Ion Penetrability Class']), model=model_name)
    r2 = r2_score(test_data[label], y_pred)
    rmse = np.sqrt(mean_squared_error(test_data[label], y_pred))
    mae = mean_absolute_error(test_data[label], y_pred)
    metrics.append({'Model': model_name, 'R2': r2, 'RMSE': rmse, 'MAE': mae})

# Convert the metrics list to a DataFrame for better visualization
metrics_df = pd.DataFrame(metrics)

# Print the metrics DataFrame
print(metrics_df)
```

```python
import pandas as pd
from autogluon.tabular import TabularPredictor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import numpy as np

# Load the original and synthetic datasets
dataset = pd.read_excel(r'cleaned_dataset_with_iso_forest.xlsx', 'Sheet1', header=0)
df_gc = pd.read_excel(r'synthetic_data_High_Low_moderate.xlsx')

# Step 1: Split the original data into training and testing sets
train_data, test_data = train_test_split(dataset, test_size=0.2, random_state=42)

# Concatenate training data with the synthetic data
train_data = pd.concat([train_data, df_gc], axis=0)

# Drop unwanted columns (columns not needed for model training)
features_to_drop = ['COULOMB_TEST', 'FA%', 'SS%', 'SF%', 'Chloride Ion Penetrability Class']

# Train the model using a regression preset
predictor = TabularPredictor(label='log', problem_type='regression').fit(
    train_data.drop(columns=features_to_drop), 
    presets='best_quality'
)

# Get a list of all model names from the leaderboard
all_models = predictor.get_model_names()

# Collect metrics for each model
metrics = []
for model_name in all_models:
    # Drop the label column and irrelevant columns from the test data
    y_pred = predictor.predict(test_data.drop(columns=['log'] + features_to_drop), model=model_name)
    r2 = r2_score(test_data['log'], y_pred)
    rmse = np.sqrt(mean_squared_error(test_data['log'], y_pred))
    mae = mean_absolute_error(test_data['log'], y_pred)
    metrics.append({'Model': model_name, 'R2': r2, 'RMSE': rmse, 'MAE': mae})

# Convert the metrics list to a DataFrame for better visualization
metrics_df1 = pd.DataFrame(metrics)

# Print the metrics DataFrame
print(metrics_df1)
```
