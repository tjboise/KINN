```python
from sdv.metadata import SingleTableMetadata  # or the new Metadata class if available
from sdv.single_table import CTGANSynthesizer
from sdv.sampling import Condition
import pandas as pd

# Step 1: Load your dataset
file_path = 'cleaned_dataset_with_iso_forest.xlsx'  # Replace with your actual file path
df = pd.read_excel(file_path, sheet_name='Sheet1')

# Step 2: Select the columns to use (including the category column)
real_data = df[['COULOMB_TEST', 'AGE', 'PC1', 'PC2', 'PC1_TYPE', 'FA', 'SS', 'SF', 'FAGG', 'CAGG', 'WATER', 
                'AEA', 'WR_HR', 'WR', 'ACC', 'VOID', 'FIBER', 'SLUMP', 'IS_PUMP', 'IS_LATEX', 
                'LATEX_AMOUNT', 'IS_LIGHT_WEIGHT', 'APPLICATION_ID', 'TOTAL_BINDER', 'TOTAL_AGG', 
                'w/b', 'b/a', 'SCM%', 'FAGG%', 'CAGG%', 'FA%', 'SS%', 'SF%', 'Chloride Ion Penetrability Class']]

# Step 3: Define metadata for the dataset
metadata = SingleTableMetadata()  # You can switch to Metadata if SDV has updated
metadata.detect_from_dataframe(data=real_data)

# Step 4: Create the CTGAN synthesizer
synthesizer = CTGANSynthesizer(metadata)

# Step 5: Train the CTGAN synthesizer on the real data
synthesizer.fit(real_data)

# Step 6: Define the conditions using the Condition class
high_condition = Condition(
    num_rows=650,  # Number of rows to generate
    column_values={'Chloride Ion Penetrability Class': 'High'}  # The exact value you want to fix
)

low_condition = Condition(
    num_rows=409,  # Number of rows to generate
    column_values={'Chloride Ion Penetrability Class': 'Moderate'}  # Another exact value for generating data
)

moderate_condition = Condition(
    num_rows=559,  # Number of rows to generate
    column_values={'Chloride Ion Penetrability Class': 'Low'}  # Another exact value for generating data
)

# Step 7: Generate synthetic data based on the defined conditions
synthetic_data = synthesizer.sample_from_conditions(
    conditions=[high_condition, moderate_condition, low_condition],  # You can add more conditions as needed
    batch_size=200,  # Batch size for sampling (adjust based on your needs)
    output_file_path=None  # Remove this if you want to save directly to a CSV
)

# Step 8: Save the synthetic data to an Excel file
synthetic_data.to_excel('synthetic_data_High_Low_moderate.xlsx', index=False)

print("Synthetic data generated successfully based on the specified conditions.")
```
