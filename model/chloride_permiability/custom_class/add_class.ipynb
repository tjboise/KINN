## add_class

```python
import pandas as pd

# Load your dataset
df = pd.read_excel('chloride_Classification.xlsx', sheet_name='Sheet1')

# Function to classify based on the Charge Passed
def classify_charge(coulombs):
    if coulombs > 2000:
        return 'High', 1
    elif 1200 <= coulombs <= 2000:
        return 'Moderate', 2
    elif 800 <= coulombs < 1200:
        return 'Low', 3
    else:
        return 'Very Low', 4

# Apply the function to the 'Charge Passed' column and create a new column 'Class'
df[['Chloride Ion Penetrability', 'Class']] = df['COULOMB_TEST'].apply(lambda x: pd.Series(classify_charge(x)))

# Save the new dataframe to a file
df.to_excel('classified_chloride_Classification.xlsx', index=False)
```

## plot bar

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
file_path = 'classified_chloride_Classification.xlsx'  # Replace with your actual file path
df = pd.read_excel(file_path, sheet_name='Sheet1')

# Specify the column you want to plot
column_name = 'Chloride Ion Penetrability'  # Replace with your actual column name

# Calculate the value counts of the column for plotting
value_counts = df[column_name].value_counts()

# Create a bar plot
# plt.figure(figsize=(10, 6))
bars = value_counts.plot(kind='bar', color='skyblue', edgecolor='black')

# Add count numbers on top of each bar
for index, value in enumerate(value_counts):
    plt.text(index, value + 7, str(value), ha='center', fontsize=10)

# Set titles and labels
# plt.title(f'Bar Plot of {column_name}')
plt.xlabel('Categories')
plt.ylabel('Count')
# plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=0)
plt.savefig('count_bar.png', dpi=300, bbox_inches='tight')

# Show the plot
# plt.show()
```
