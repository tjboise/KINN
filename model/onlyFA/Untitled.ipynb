```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the PDP results from the CSV file
pdp_results = pd.read_csv('pdp_results_28.csv')  # Change the path if necessary

# Define colors similar to seaborn's vlag palette
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']  # Blue, Orange, Green, Red, Purple, Brown

# Set the figure size
# plt.figure(figsize=(12, 8))

# Plot each model's predictions with specified colors
plt.plot(pdp_results['FA% (Original Scale)'], pdp_results['XGBoost'], label='XGBoost', color=colors[0])
plt.plot(pdp_results['FA% (Original Scale)'], pdp_results['ANN'], label='ANN', color=colors[1])
plt.plot(pdp_results['FA% (Original Scale)'], pdp_results['RF'], label='Random Forest', color=colors[2])
plt.plot(pdp_results['FA% (Original Scale)'], pdp_results['SVR'], label='SVR', color=colors[3])
plt.plot(pdp_results['FA% (Original Scale)'], pdp_results['KINN'], label='KINN', color=colors[4])
plt.plot(pdp_results['FA% (Original Scale)'], pdp_results['SKIRM'], label='SKIRM', color=colors[5])

# Adding titles and labels
# plt.title('Partial Dependence of fc on FA% for Different Models')
plt.xlabel('FA%')
plt.ylabel('fc (MPa)')
# plt.axhline(0, color='black', linestyle='--', label='Zero Line')  # Optional: Add zero line for reference
plt.legend()
# plt.grid()
# plt.show()
plt.xlim(0,0.6)
plt.tight_layout()

# Save the figure with better resolution and spacing
plt.savefig('PDP_28.png', dpi=300)
```
