```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load and preprocess the data
df = pd.read_excel(r'updated_fc_predictions_with_fc_hat.xlsx', sheet_name='7')
df.dropna(inplace=True)

# Define features
features = ['PC', 'PC_TYPE', 'FA',  'FAGG', 'CAGG', 'WATER', 'AEA', 'WR_HR', 'WR', 'ACC', 
            'VOID', 'w/b', 'b/a', 'CAGG%', 'FAGG%', 'FA%']

# Original values for features, excluding 'w/b'
mean_values = df[features].mean(axis=0)

# Define constants for the empirical formula
a = 40.5
b = 15.3
e = 6.5
d = 0.36

# Create a range of 'w/b' values for the PDP
wb_range = np.linspace(df['w/b'].min(), df['w/b'].max(), 100)

# Prepare a DataFrame for predictions, keeping other features fixed at their mean values
pdp_data = pd.DataFrame(np.tile(mean_values, (wb_range.size, 1)), columns=features)
pdp_data['w/b'] = wb_range  # Vary 'w/b' over the range

# Calculate predicted 'fc' using the empirical formula for each 'w/b'
AGE = 7  # Set AGE to 28
fc_pred = (a * np.log(AGE) + b) * (e * AGE ** d) ** (-pdp_data['w/b'].values)

# Save predictions to a DataFrame
pdp_results = pd.DataFrame({
    'w/b': wb_range, 
    'Predicted_fc': fc_pred
})

# Save the predictions to a CSV file
pdp_results.to_csv('pdp_EE_wb_7.csv', index=False)

# Plot the Partial Dependence Plot (PDP)
plt.figure(figsize=(8, 6))
plt.plot(pdp_results['w/b'], pdp_results['Predicted_fc'], label='Predicted fc', color='blue')
plt.xlabel('w/b', fontsize=14)
plt.ylabel('Predicted fc (MPa)', fontsize=14)
plt.title('Partial Dependence Plot: Predicted fc vs w/b', fontsize=16)
plt.grid(True)
plt.show()
```
