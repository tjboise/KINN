```python
import pandas as pd
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# Load the updated dataset
df = pd.read_excel(r'updated_fc_predictions.xlsx', sheet_name='Sheet1')
df.dropna(inplace=True)

# Define features and target
X = df[['AGE', 'PC', 'PC_TYPE', 'FA', 'SS', 'SF', 'FAGG', 'CAGG', 'WATER', 'AEA', 'WR_HR', 'WR', 'ACC', 
        'VOID', 'w/b', 'b/a', 'CAGG%', 'FAGG%', 'FA%', 'SS%', 'SF%']].values

y = df['fc (MPa)'].values

# Define the curve fitting model function
def fc_model(X, a, b, e, d):
    AGE = X[:, 0]
    wb = X[:, -7]  # w/b should be the last feature (index -1)
    
    # Equation: fc = (a * log(AGE) + b) * (e * AGE^d)^(-wb)
    fc_pred = (a * np.log(AGE) + b) * (e * AGE ** d) ** (-wb)
    
    return fc_pred

# Initialize a DataFrame to store results
results_df = pd.DataFrame(columns=['Run', 'Test R2', 'Train R2', 'RMSE', 'MAE'])

# Define a function to run the model and calculate metrics
for run in range(1, 101):
    print(f'Run {run}')
    
    # Split the data into training and testing sets
    Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, train_size=0.85, random_state=run)
    
    # Initial guess for the parameters
    initial_guess = [1, 1, 1, 1]
    
    # Fit the model using curve_fit on the training data
    popt, _ = curve_fit(fc_model, Xtrain, ytrain, p0=initial_guess)
    
    # Extract fitted parameters
    a_fit, b_fit, e_fit, d_fit = popt
    
    # Predictions for the train and test datasets
    y_train_pred = fc_model(Xtrain, a_fit, b_fit, e_fit, d_fit)
    y_test_pred = fc_model(Xtest, a_fit, b_fit, e_fit, d_fit)
    
    # Calculate R² for train and test datasets
    train_r2 = r2_score(ytrain, y_train_pred)
    test_r2 = r2_score(ytest, y_test_pred)
    
    # Calculate RMSE and MAE for both train and test datasets
    rmse = np.sqrt(mean_squared_error(ytest, y_test_pred))
    mae = mean_absolute_error(ytest, y_test_pred)
    
    # Append the results to the DataFrame
    results_df = results_df.append({
        'Run': run,
        'Test R2': test_r2,
        'Train R2': train_r2,
        'RMSE': rmse,
        'MAE': mae
    }, ignore_index=True)
    
    # Print the results for this run
    print(f'Run {run}: Test R² = {test_r2}, Train R² = {train_r2}, RMSE = {rmse}, MAE = {mae}')

# Save the results to an Excel file
results_df.to_excel('empirical_model_results.xlsx', index=False)

print("Results and models saved successfully.")
```

```python
import pandas as pd
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.preprocessing import MinMaxScaler

# Load the updated dataset
df = pd.read_excel(r'updated_fc_predictions.xlsx', sheet_name='Sheet1')
df.dropna(inplace=True)

# Define features and target
X = df[['AGE', 'PC', 'PC_TYPE', 'FA', 'SS', 'SF', 'FAGG', 'CAGG', 'WATER', 'AEA', 'WR_HR', 'WR', 'ACC', 
        'VOID', 'w/b', 'b/a', 'CAGG%', 'FAGG%', 'FA%', 'SS%', 'SF%']].values

y = df['fc (MPa)'].values

# Scale the features and target variable to a range of [0, 1]
feature_scaler = MinMaxScaler()
X_scaled = feature_scaler.fit_transform(X)

target_scaler = MinMaxScaler()
y_scaled = target_scaler.fit_transform(y.reshape(-1, 1)).flatten()

# Define the curve fitting model function
def fc_model(X, a, b, e, d):
    AGE = X[:, 0]
    wb = X[:, -7]  # w/b should be the last feature (index -1)
    
    # Equation: fc = (a * log(AGE) + b) * (e * AGE^d)^(-wb)
    fc_pred = (a * np.log(AGE) + b) * (e * AGE ** d) ** (-wb)
    
    return fc_pred

# Initialize a DataFrame to store results
results_df = pd.DataFrame(columns=['Run', 'Test R2', 'Train R2', 'RMSE', 'MAE'])

# Define a function to run the model and calculate metrics
for run in range(1, 101):
    print(f'Run {run}')
    
    # Split the data into training and testing sets
    Xtrain, Xtest, ytrain, ytest = train_test_split(X_scaled, y_scaled, train_size=0.85, random_state=run)
    
    # Initial guess for the parameters
    initial_guess = [1, 1, 1, 1]
    
    # Fit the model using curve_fit on the training data
    popt, _ = curve_fit(fc_model, Xtrain, ytrain, p0=initial_guess)
    
    # Extract fitted parameters
    a_fit, b_fit, e_fit, d_fit = popt
    
    # Predictions for the train and test datasets (scaled)
    y_train_pred = fc_model(Xtrain, a_fit, b_fit, e_fit, d_fit)
    y_test_pred = fc_model(Xtest, a_fit, b_fit, e_fit, d_fit)
    
    # Reverse the scaling of the predictions
    y_train_pred_orig = target_scaler.inverse_transform(y_train_pred.reshape(-1, 1)).flatten()
    y_test_pred_orig = target_scaler.inverse_transform(y_test_pred.reshape(-1, 1)).flatten()
    
    # Reverse the scaling of the true values
    ytrain_orig = target_scaler.inverse_transform(ytrain.reshape(-1, 1)).flatten()
    ytest_orig = target_scaler.inverse_transform(ytest.reshape(-1, 1)).flatten()
    
    # Calculate R² for train and test datasets
    train_r2 = r2_score(ytrain_orig, y_train_pred_orig)
    test_r2 = r2_score(ytest_orig, y_test_pred_orig)
    
    # Calculate RMSE and MAE for both train and test datasets
    rmse = np.sqrt(mean_squared_error(ytest_orig, y_test_pred_orig))
    mae = mean_absolute_error(ytest_orig, y_test_pred_orig)
    
    # Append the results to the DataFrame
    results_df = results_df.append({
        'Run': run,
        'Test R2': test_r2,
        'Train R2': train_r2,
        'RMSE': rmse,
        'MAE': mae
    }, ignore_index=True)
    
    # Print the results for this run
    print(f'Run {run}: Test R² = {test_r2}, Train R² = {train_r2}, RMSE = {rmse}, MAE = {mae}')

# Save the results to an Excel file
results_df.to_excel('empirical_model_results_scaled_features.xlsx', index=False)

print("Results and models saved successfully.")
```
